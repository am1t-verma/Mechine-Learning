17) Which column has the highest entropy excluding the target column Outcome in the original data?

```python
df
```

```python
col = df.columns
```

```python
coll=[]
for i in col:
    coll.append(i)
```

```python
coll
```

```python
coll.remove('Outcome')
```

```python
coll
```

```python
ent_list = []
col_list = []

for i in coll:
    probs = df[i].value_counts(normalize = True)

    entropy = (-1*(np.sum(np.log2(probs) * probs)))
    col_list.append(i)
    ent_list.append(entropy)
```

```python
col_list
```

```python
pd.DataFrame({'entropy' : ent_list, 'features' : col_list}).sort_values(by = 'entropy', ascending=False)
```

18) Which column is considered as the root node of the decision tree using Information Gain in the original data?


```python
df.info()
```

```python
import numpy as np

gini_dict = {}

for col in df.columns:
    probs = df[col].value_counts(normalize=True)
    gini = 1 - np.sum(probs**2)
    gini_dict[col] = gini

print(gini_dict)
```

```python
pd.DataFrame({'gine':gini_dict}).sort_values(by='gine')

```
