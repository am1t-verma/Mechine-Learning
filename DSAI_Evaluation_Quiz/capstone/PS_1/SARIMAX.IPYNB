```python
import numpy as np
import pandas as pd
import seaborn  as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Walmart.csv")
df
```

```python
df.info()
```

```python
df['Date'] = pd.to_datetime(df['Date'], format= "%d-%m-%Y")
```

```python
store1 = df[df['Store'] == 1]

```

```python
store1.drop(columns = ['Store', 'Holiday_Flag', 'Temperature',
       'Fuel_Price', 'CPI', 'Unemployment'], inplace = True)
```

```python
store1.info()
```

```python
sns.lineplot(store1, x = 'Date', y = 'Weekly_Sales')
```

```python
from statsmodels.tsa.seasonal import seasonal_decompose
decompose = seasonal_decompose(store1['Weekly_Sales'],model='additive',period=12)
decompose.plot()
plt.show()
```

```python
rolling_mean = store1['Weekly_Sales'].rolling(window=3).mean()  # 52 weeks = 1 year
rolling_std  = store1['Weekly_Sales'].rolling(window=3).std()

plt.figure(figsize=(12,5))
plt.plot(store1['Weekly_Sales'], color='blue',  label='Original')
plt.plot(rolling_mean,         color='red',   label='Rolling Mean')
plt.plot(rolling_std,          color='black', label='Rolling Std')
plt.legend()
plt.title('Rolling Mean & Std')
plt.show()
```

```python
from statsmodels.tsa.stattools import adfuller
result = adfuller(df['Weekly_Sales'])

result[1]
```

```python
if result[1] > .05:
    print('Non-Stationar')

else:
    print('Stationary')
```

```python
from statsmodels.tsa.stattools import acf
from statsmodels.graphics.tsaplots import plot_acf


acf_values = acf(store1['Weekly_Sales'], nlags=20)
print(acf_values)
plot_acf(store1['Weekly_Sales'], lags=10)
plt.show()
```

```python
from statsmodels.graphics.tsaplots import plot_pacf

plot_pacf(store1['Weekly_Sales'], lags=10)
plt.show()
```

```python
train = store1['Weekly_Sales'].iloc[:115]
test = store1['Weekly_Sales'].iloc[115:]
```

```python
from statsmodels.tsa.statespace.sarimax import SARIMAX

model = SARIMAX(train, order = (1, 1, 1), seasonal_order = (1, 1, 1, 4))

model_fit = model.fit(disp=False)
print(model_fit.summary())
```

```python
pred = model_fit.predict(start = len(train), 
                         end = len(train) + len(test) - 1,
                         dynamic = False)
```

```python
plt.figure(figsize=(10,6))
plt.plot(train, label = 'Train', color = 'blue')
plt.plot(test, label = 'Test', color = 'green')
plt.plot(pred, label = 'Prediction', color = 'red')
plt.legend()
plt.show()
```

```python
forcast = model_fit.forecast(steps=len(test))
forcast.index = test.index

plt.figure(figsize=(10,6))
plt.plot(train, label = 'Train', color = 'blue')
plt.plot(test, label = 'Test', color = 'green')
plt.plot(forcast, label = 'forcast', color = 'red')
plt.legend()
plt.show()
```
