```python
store = {}   # ← define dictionary first

for i in df['Store'].unique():
    store[i] = df[df['Store'] == i].copy()
    store[i].drop(columns=['Store'], inplace=True)
    store[i] = store[i].reset_index(drop=True)  # ← save it back
    
    model = Prophet(
        yearly_seasonality=True,
        weekly_seasonality=False,
        daily_seasonality=False
    )
    model_fit = model.fit(store[i]) # here we put store1 like that
    future = model.make_future_dataframe(periods = 84)
    forecast = model.predict(future)
    
    model.plot(forecast)
    plt.savefig(f'store_plots/store_{i}_forecast.png',
                bbox_inches='tight',                     
                dpi=150)
    plt.legend()
    plt.title(f'Store No. {i}')
    plt.xlabel('Weeks')
    plt.ylabel('Weekly Sales')
    plt.show()
    plt.close()
    print(f"Store {i} saved ✅")
```
