```python
import matplotlib.dates as mdates
plt.figure(figsize=(10, 6))
sns.lineplot(df, x='Date', y = 'Weekly_Sales')
plt.gca().xaxis.set_major_locator(mdates.MonthLocator(interval=1))
plt.xticks(rotation = 90)
plt.tight_layout()
plt.grid()
plt.show()
```
